## Figures generation

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import math
import subprocess
import os

%matplotlib inline

plt.style.use('classic')

plt.rcParams['figure.facecolor']='white'
plt.rcParams['figure.figsize'] = (8.0, 5.0)

img_jacketed=mpimg.imread('0To1J_Clifton.png')
img_unjacketed=mpimg.imread('1To0U_Clifton.png')
```

```python



plt.style.use('classic')

plt.rcParams['figure.facecolor']='white'
plt.rcParams['figure.dpi']= 150

g_theta_dir = 'G_theta/'
plt.imshow(img_jacketed, extent=[0,1,0,1.2])
w3_r0 = g_theta_dir+'W_3/G_W3_r0.out'
w4_r0 = g_theta_dir+'W_4/G_W4_r0.out'
w7_r0 = g_theta_dir+'W_7/G_W7_r0.out'
w10_r0 = g_theta_dir+'W_10/G_W10_r0.out'

ell_w3 = []
G_w3 = []
with open(w3_r0) as inputfile:
    for line in inputfile:
        line = line.split(',')
        ell_w3.append(float(line[0]))
        G_w3.append(float(line[1]))

ell_w4 = []
G_w4 = []
with open(w4_r0) as inputfile:
    for line in inputfile:
        line = line.split(',')
        ell_w4.append(float(line[0]))
        G_w4.append(float(line[1]))
        
ell_w7 = []
G_w7 = []
with open(w7_r0) as inputfile:
    for line in inputfile:
        line = line.split(',')
        ell_w7.append(float(line[0]))
        G_w7.append(float(line[1]))  

ell_w10 = []
G_w10 = []
with open(w10_r0) as inputfile:
    for line in inputfile:
        line = line.split(',')
        ell_w10.append(float(line[0]))
        G_w10.append(float(line[1]))  
        
plt.xlim(0,1)
plt.ylim(0,1.2)
plt.grid()
plt.xlabel('$l$',fontsize=18)
plt.ylabel('$K_I^{J*}$',fontsize=18)
plt.plot(ell_w3, G_w3 ,'b', label='W_3',marker = '', linestyle='dashed',)
plt.plot(ell_w4, G_w4 ,'r', label='W_4',marker = '', linestyle='dashed',)
plt.plot(ell_w7, G_w7 ,'g', label='W_7',marker = '', linestyle='dashed',)
plt.plot(ell_w10, G_w10 ,'y', label='W_10',marker = '', linestyle='dashed',)
plt.savefig('Clifton_Comparison_J.png')
```

```python
plt.rcParams['figure.facecolor']='white'
plt.rcParams['figure.dpi']= 150

w12_r0 = g_theta_dir+'/W_12/G_W12_r0.out'
w12_c = g_theta_dir+'/W_12/G_W12_conf.out'


ell_w12 = []
G_J_w12 = []
with open(w12_r0) as inputfile:
    for line in inputfile:
        line = line.split(',')
        ell_w12.append(float(line[0]))
        G_J_w12.append(float(line[1]))

G_C_w12 = []
with open(w12_c) as inputfile:
    for line in inputfile:
        line = line.split(',')
        G_C_w12.append(float(line[1]))
         
plt.xlabel('$\ell$',fontsize=18)
plt.ylabel('$K_I^{B*}$ $(w=12)$',fontsize=18)
plt.xlim(0,0.9)
plt.ylim(-0.5,1.0)
plt.plot(ell_w12, G_J_w12 ,'k', label='$p^* = 0$',marker = '', linestyle='solid',)
plt.plot(ell_w12, np.array(G_J_w12[:]) - 1/8.*np.array(G_C_w12[:]) ,'k', label='$p^* = 1/8$',marker = '', linestyle='dotted',)
plt.plot(ell_w12, np.array(G_J_w12[:]) - 1/6.*np.array(G_C_w12[:]) ,'k', label='$p^* = 1/6$',marker = '', linestyle='dashed',)


plt.legend(loc='upper left',ncol=1,fontsize=10)
plt.savefig('W12_various.png')
```
